```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]

y = [1,4,9,16,25]

#Creating  a graph
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title("Basic Line plot")
plt.show()
```

```python
x = [1,2,3,4,5]

y = [1,4,9,16,25]

#Creating a Customized Line plot
# plt.plot(x,y,color='red',linestyle='-.')
# plt.plot(x,y,color='red',linestyle='-.',marker='o')
# plt.plot(x,y,color='red',linestyle='-.',marker='o',linewidth=2)
plt.plot(x,y,color='red',linestyle='-.',marker='o',linewidth=2,markersize=9)
plt.grid(True)
```

```python
# Multiple plots

x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

plt.figure(figsize=(9,5))
plt.subplot(2,2,1) # 2 x 2 matrix where 4 graph can accamodate the third parameter is the position of the graph
plt.plot(x,y1,color='green')


plt.subplot(2,2,2) # 2 x 2 matrix where 4 graph can accamodate the third parameter is the position of the graph
plt.plot(y1,x,color='red')


plt.subplot(2,2,3) # 2 x 2 matrix where 4 graph can accamodate the third parameter is the position of the graph
plt.plot(y1,y2,color='blue')

plt.subplot(2,2,4) # 2 x 2 matrix where 4 graph can accamodate the third parameter is the position of the graph
plt.plot(y2,x,color='yellow')
```

```python
# Bar Plot
categories = ['A','B','C','D','E']
values = [5,6,7,8,6]

plt.bar(categories,values,color='purple')
```

## Histograms

Histograms are used to represent the distribution of a dataset. They divide the data into bins and count the number of data points in each bin

```python
data = [1,2,2,3,3,3,4,4,4,5,5,5,5]

plt.hist(data,bins=5,color='orange',edgecolor='black')
```

```python
#creating a scatter plot

x = [1,2,3,4,5]
y = [2,3,4,5,6]

plt.scatter(x,y,color="blue",marker='x')
```

```python
#pie chart

labels = ['A','B','C','D']
sizes = [30,20,40,10]
colors = ['gold','yellowgreen','lightcoral','pink']
explode = (0.1,0,0,0) #Moves out the first slice out the parameter mentioned is the distance from the center point

plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct="%1.1f%%",shadow=True)
```
