```python
import sqlite3
```

```python
# Connect to an SQLite database
connection = sqlite3.connect('example.db')
```

```python
cursor = connection.cursor()
```

```python
# Create a Table
cursor.execute('''
Create Table If Not Exists employees(
        id Integer Primary Key,
        name Text Not Null,
        age Integer,
        department text
        )
''')


# Commit the changes
connection.commit()
```

```python
cursor.execute('''
    Select * from employees
''')
```

```python
# INsert the data in sqlite table

cursor.execute('''
    Insert into employees(name,age,department)
               values('Gopal',22,'Berosgaar')
''')

cursor.execute('''
    Insert into employees(name,age,department)
               values('Ram',342,'Busy')
''')

cursor.execute('''
    Insert into employees(name,age,department)
               values('Sham',562,'Bhot Busy')
''')

connection.commit()
```

```python

cursor.execute('select * from employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# Update data in the table

cursor.execute('''
    UPDATE employees
    set age = 23
    where name = 'Gopal'
''')
```

```python
# Delete the data from the table

cursor.execute('''
        delete from employees 
                where name = 'Ram'               
   ''')
```

```python
data = [
    (7,'Unknown', 324,'Player'),
    (8,'Unknown1', 324,'Proo player'),
    (9,'Unknown2', 324,'noob Player')
]

cursor.executemany('''
    Insert into employees values(?,?,?,?)
''',data)

# connection.commit()
```

```python

cursor.execute('select * from employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```
