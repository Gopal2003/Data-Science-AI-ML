```python
arr = np.arange(25).reshape(5,5)

# print(arr)


print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.var(arr))
```

```python
arr = np.arange(1,10).reshape(3,3)
print(arr)

mean_val = np.mean(arr)
std_val = np.std(arr)

# Step 3: Normalize
normalized_arr = (arr - mean_val) / std_val

print(normalized_arr)
print(np.mean(normalized_arr))
```

```python
person_dtype = np.dtype([
    ('name', 'U20'),   # string of up to 20 characters
    ('age', 'i4'),     # 32-bit integer
    ('weight', 'f4')   # 32-bit float
])

arr = np.array([

    ('Gopal',23,66.30),
    ('Krishna',27,42.34),
    ('Ram',21,77.35)

],dtype=person_dtype)


# print(arr)

sorted_people = np.sort(arr, order = 'age')

print(sorted_people)
```
