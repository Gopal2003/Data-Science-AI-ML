```python
arr2 = np.arange(25).reshape(5,5)

print(arr2,"\n\n")
print(arr2[0:1,0:],"\n") # Top Border
print(arr2[1:,0:1],"\n") # Left Border
print(arr2[4:5,1:],"\n") # Bottom border
print(arr2[1:4,4:5],"\n") # Right border

```
