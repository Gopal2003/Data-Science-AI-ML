## Data Manipulation using Pandas


```python
df = pd.read_csv('data.csv')

# To fetch the first and last few rows
print(df.head(5), "\n")
print(df.tail(5))
```

```python
df.describe()
```

```python
df.dtypes
```

```python
# Hanlding Missing Values

df.isnull().any(axis = 1)
```

```python
df.isnull().sum()
```

```python
df.fillna(0)
```

```python
# Fill the Missing Values in Calories with the mean of the Calories.

df.fillna(df['Calories'].mean())
```

```python
# Fill the Missing Values of Calories with the Mean of Calories into a new Column

df['Sales_fillna'] = df['Calories'].fillna(df['Calories'].mean())

df
```

```python
# Renaming Columns
df = df.rename(columns={'Maxpulse':'Max-Pulse'})
df.head()
```

```python
df.dtypes
```

```python

#Changing the Data type of the column

df['Calories_New'] = df['Calories'].fillna(df['Calories'].mean()).astype(int)

df
```

```python
# Apply a function

df['Pulse*2'] = df['Pulse'].apply(lambda x:x * 2)
df
```
