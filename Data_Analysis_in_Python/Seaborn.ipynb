```python
import seaborn as sns
```

```python
# Basic Plotting with Seaborn

tips = sns.load_dataset('tips')

tips

```

```python
#Create a Scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x = 'total_bill',y='tip',data=tips)
plt.title("Scatter Plot of Total Bills vs Tip")
plt.show()
```

```python
#Line Plot

sns.lineplot(x='sex',y='size',data=tips)
plt.title("Line plot graph sex vs size")
```

```python
#Categorical Plots

# Bar Plot

sns.barplot(x='day',y='total_bill',data=tips)
plt.title('Bar Plot of total bill by dat')
plt.show()
```

```python
#Box Plot
sns.boxplot(x="day",y="total_bill",data=tips)
```

```python
tips
```

```python
# Violin Plot

sns.violinplot(x='day',y="total_bill",data=tips)
```

```python
#Histograms
sns.histplot(tips['total_bill'],bins=10,kde=False)
```

```python
# KDE plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
# Pairplot
# Shows the relationship between each column
sns.pairplot(tips)
```

```python
# HeatMap
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```
