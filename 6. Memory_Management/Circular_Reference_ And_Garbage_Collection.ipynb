```python
import gc
import sys

class Myobject:
    def __init__(self,name):
        self.name = name

    def __del__(self):
        print(f"Object {self.name} Deleted!")


obj1 = Myobject("Object 1")
obj2 = Myobject("Object 2")
obj1.ref = obj2
obj2.ref = obj1

# print(sys.getrefcount(obj2))
# print(sys.getrefcount(obj1))

del obj1
del obj2

# print(sys.getrefcount(obj2))
# print(sys.getrefcount(obj1))

# Manually Triggring the Garbage collector because of circular Reference
# gc.collect()
print(gc.garbage)

# Reason for multiple deleteed print statement:

# 🧩 Step-by-step explanation
# 1️⃣ First run (without gc.collect())

# When you executed:

# del obj1
# del obj2


# you removed the names obj1 and obj2, but since they referenced each other:

# obj1.ref = obj2
# obj2.ref = obj1


# they formed a circular reference, so their reference counts never dropped to 0.

# Hence:

# The garbage collector didn’t immediately destroy them.

# __del__() was not called.

# The objects stayed in memory (uncollected, in a hidden cycle).

# 2️⃣ Second run (after uncommenting gc.collect())

# Now, you reran the script in the same Python session (like inside IDLE, VS Code, or Jupyter).

# That means:

# The old uncollected objects from the previous run were still hanging around in memory (in the garbage collector’s cyclic buffer).

# When you executed:

# gc.collect()


# the garbage collector finally cleaned up those old cycle objects, calling their __del__() methods for the first time.

# Then your script created new objects (obj1 and obj2 again), deleted them, and the GC ran again — calling __del__() a second time for these new objects.

# Hence you saw:

# Object Object 1 Deleted!
# Object Object 2 Deleted!
# Object Object 1 Deleted!

```
