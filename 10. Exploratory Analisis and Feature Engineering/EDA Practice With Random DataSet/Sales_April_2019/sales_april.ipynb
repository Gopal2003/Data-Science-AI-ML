```python
# Converting the Date,month,year format into separate columns
data_copy[['Date','Month','Year']] = data_copy['Order Date'].str.split('/',expand=True)
```

```python
data_copy[['Year','Time']] = data_copy['Year'].str.split(' ',expand=True)
```

```python
removed_rows = data_copy[data_copy['Date'] == 'Order Date']
# data_copy = da

removed_rows
```

```python
data_copy = data_copy[data_copy['Date'] != 'Order Date'].reset_index(drop=True)
# data_copy.shape

```

```python
data_copy['Date'] = data_copy['Date'].astype('int')
data_copy['Month'] = data_copy['Month'].astype('int')
data_copy['Year'] = data_copy['Year'].astype('int')
# data_copy['Date']

data_copy
```

```python
data_copy.info()
```

```python
data_copy = data_copy.drop('Order Date',axis=1)
```

```python
data_copy
# removed_rows
```

```python
data_copy['Hours'] = data_copy['Time'].str.split(':').str[0]
data_copy['Minutes'] = data_copy['Time'].str.split(':').str[1]
```

```python
data_copy = data_copy.drop('Time',axis=1)
```

```python
data_copy['Hours'] = data_copy['Hours'].astype('int')
data_copy['Minutes'] = data_copy['Minutes'].astype('int')
```
