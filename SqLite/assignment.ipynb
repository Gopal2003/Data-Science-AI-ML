```python
import sqlite3
```

```python
connection = sqlite3.connect('test.db')
cursor = connection.cursor()
```

```python
cursor.execute('''
    create table if not exists employees(
        id Integer,
        name text,
        age Integer,
        department text
               )
''')
```

```python
cursor.execute('''
    select * from employees
''')
```

```python
data = [
        (101,'Gopal',22,'Technical'),
        (102,'Ram',23,'Civil'),
        (103,'Sham',24,'Mechanical')
]

cursor.executemany('''
        insert into employees values(?,?,?,?)
''',data)
```

```python
cursor.execute('select * from employees')
rows = cursor.fetchall()


for row in rows:
    print(row)
```

```python
def run_query(query):
    cursor.execute(query)
    rows = cursor.fetchall()
    for row in rows:
        print(row)
```

```python
run_query("select * from employees where department = 'Technical'")
```

```python
run_query("update employees set department = 'Technical' " \
"where id = 103")
```

```python
run_query()
```
